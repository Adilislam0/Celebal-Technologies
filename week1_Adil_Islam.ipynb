### 1.3 Exceptions

Write `safe_divide(a, b)` that:
- Returns `a / b` normally
- Returns `None` if `b == 0` (catch `ZeroDivisionError`)
- Raises `TypeError` with message `"Inputs must be numeric"` if either input is not a number

````python
def safe_divide(a, b):
    # Check types first — both inputs must be numeric (int or float)
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Inputs must be numeric")
    try:
        return a / b
    except ZeroDivisionError:
        return None

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```
A = [[4, 1],
     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v   = eigenvectors[:, i]
    Av  = A @ v
    lv  = lam * v
    match = np.allclose(Av, lv)
    print(f"  Eigenpair {i}: λ={lam:.2f}, Av≈λv? {match}")

# Plot eigenvectors scaled by eigenvalues
fig, ax = plt.subplots()
colors = ['royalblue', 'tomato']
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i] * eigenvalues[i]  # scale by eigenvalue
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.05, label=f'λ={eigenvalues[i]:.1f}')
ax.set_xlim(-5, 5); ax.set_ylim(-5, 5)
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.set_title('Eigenvectors scaled by Eigenvalues')
ax.legend(); plt.tight_layout(); plt.show()

print("4.3 passed ✓")
````

### Geometric Meaning of Eigenvalues/Eigenvectors

When a matrix transforms a vector, most vectors change **both** their direction and magnitude. Eigenvectors are special — they only get **stretched or flipped**, never rotated. The eigenvalue λ tells you *how much* the stretch is: λ > 1 means the vector gets longer in that direction, 0 < λ < 1 means it shrinks, λ < 0 means it flips. Think of a matrix as a rubber-sheet transformation — eigenvectors are the axes along which the sheet simply stretches, and eigenvalues measure how much.


### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False)

# Reconstruct: X ≈ U @ diag(S) @ Vt
X_reconstructed = U @ np.diag(S) @ Vt

# Rank-1 approximation: use only the largest singular value/vectors
X_approx = S[0] * np.outer(U[:, 0], Vt[0, :])

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> *Explain: what are U, S, Vt in SVD, and how do they relate to principal components? Which matrix contains the directions of maximum variance?*

### SVD → PCA Connection

In SVD, **U** contains the left singular vectors (directions in the output/row space), **S** contains the singular values (measure of variance along each direction, in descending order), and **Vt** contains the right singular vectors (directions in the input/column space — these are the **principal component directions**). PCA on a centered data matrix X is mathematically equivalent to its SVD: the rows of **Vt** are the principal components, and the singular values squared (divided by n-1) are the eigenvalues of the covariance matrix, representing variance explained by each component. So **Vt** contains the directions of maximum variance.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each sample against N(0,1) CDF
ks_s1 = stats.kstest(s1, 'norm')   # s1 is normal → p should be high
ks_s2 = stats.kstest(s2, 'norm')   # s2 is exponential → p should be low
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# Non-stationary time series: linear trend + noise
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once to remove trend, re-run ADF
ts_diff  = np.diff(ts)              # first difference removes linear trend
adf_diff = adfuller(ts_diff)
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

### 5.5 Model Monitoring Concepts

1. Implement `compute_psi(expected, actual, bins=10)` that computes the **Population Stability Index** between two distributions.
2. Simulate a scenario where a model's input distribution shifts and plot both distributions.
3. In the markdown cell, define: *concept drift*, *covariate drift*, PSI thresholds (what values indicate no/minor/major shift), and one trigger for retraining.

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    epsilon = 1e-10
    # Build common bin edges from both distributions combined
    breakpoints = np.linspace(min(expected.min(), actual.min()),
                               max(expected.max(), actual.max()), bins + 1)
    
    # Count how many values fall in each bin for expected and actual
    exp_counts = np.histogram(expected, bins=breakpoints)[0]
    act_counts = np.histogram(actual,   bins=breakpoints)[0]
    
    # Convert to percentages (proportions), add epsilon to prevent log(0)
    exp_pct = exp_counts / len(expected) + epsilon
    act_pct = act_counts / len(actual)   + epsilon
    
    # PSI formula
    psi = np.sum((act_pct - exp_pct) * np.log(act_pct / exp_pct))
    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
ax.hist(train_dist, bins=30, alpha=0.5, density=True, color='steelblue', label='Training (expected)')
ax.hist(drift_dist, bins=30, alpha=0.5, density=True, color='tomato',    label='Production (actual)')
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

### Concept Drift vs Covariate Drift & PSI Thresholds

- **Concept drift:** The statistical relationship between input features X and the target y changes over time (the model's decision boundary becomes invalid even if input distribution stays the same).
- **Covariate drift:** The distribution of input features X changes while the relationship between X and y remains the same (inputs look different but the model logic is still valid).
- **PSI < 0.1:** No significant shift — the production distribution is stable; no action needed.
- **PSI 0.1–0.2:** Minor shift — monitor closely and investigate feature-level changes.
- **PSI > 0.2:** Major shift — model retraining is recommended; the input distribution has changed substantially.
- **Retraining trigger example:** Schedule automated retraining when PSI exceeds 0.2 on the top 5 most important features, checked weekly on a rolling 30-day window of production data.


### 6.1 Core Concepts

A bag contains: 4 red, 3 blue, 3 green marbles.

1. Define the sample space and compute P(red), P(blue), P(green).
2. Two marbles drawn **without replacement**. Compute:
   - P(first=red, second=blue) — **joint probability**
   - P(second=blue | first=red) — **conditional probability**
3. Are the two draws **independent**? Show mathematically.

````python
total = 10
red, blue, green = 4, 3, 3

p_red   = red / total        # 4/10
p_blue  = blue / total       # 3/10
p_green = green / total      # 3/10

# Joint: P(red first AND blue second) — without replacement
# P(red first) * P(blue second | red first drawn) = 4/10 * 3/9
p_red_then_blue = (red / total) * (blue / (total - 1))

# Conditional: P(blue second | red first already drawn)
# After drawing red, 9 marbles remain, 3 are blue
p_blue_given_red = blue / (total - 1)

# Independence check: independent iff P(B|A) == P(B)
# P(blue|red drawn) = 3/9 = 0.333 ≠ P(blue) = 3/10 = 0.3 → NOT independent
independent = abs(p_blue_given_red - p_blue) < 1e-9

print(f"P(red)={p_red:.3f}, P(blue)={p_blue:.3f}, P(green)={p_green:.3f}")
print(f"P(red,blue): {p_red_then_blue:.4f}")
print(f"P(blue|red): {p_blue_given_red:.4f}")
print(f"P(blue):     {p_blue:.4f}")
print(f"Independent? {independent}")
````

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability: P(free) = P(free|spam)*P(spam) + P(free|ham)*P(ham)
p_free = p_free_given_spam * p_spam + p_free_given_ham * p_ham

# Bayes' Theorem: P(spam|free) = P(free|spam) * P(spam) / P(free)
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # Apply Bayes' theorem: posterior = likelihood * prior / evidence
    p_ham_local  = 1 - prior_spam
    p_word       = p_word_given_spam * prior_spam + p_word_given_ham * p_ham_local
    return (p_word_given_spam * prior_spam) / p_word

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

### Bayes Term Mapping (6.3)

| Term | Formula | In this example |
|------|---------|-----------------| 
| Prior | P(Spam) | 0.30 — our initial belief that any email is spam before seeing any words |
| Likelihood | P(word \| Spam) | P("free" \| Spam) = 0.80 — how often spam emails contain the word "free" |
| Evidence | P(word) | P("free") = 0.275 — overall probability of seeing "free" in any email |
| Posterior | P(Spam \| word) | P(Spam \| "free") ≈ 0.872 — updated belief that the email is spam after seeing "free" |


````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples   = 5000
sample_size = 30

# Draw 5000 samples of size 30 and record each mean
sample_means = np.array([
    np.random.choice(population, size=sample_size, replace=False).mean()
    for _ in range(n_samples)
])

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)  # CLT: std of sample means = σ/√n

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot histogram of sample means + theoretical normal overlay
fig, ax = plt.subplots()
ax.hist(sample_means, bins=50, density=True, alpha=0.6, color='steelblue', label='Sample Means')
x_range = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(x_range, stats.norm.pdf(x_range, pop_mean, clt_std),
        color='tomato', lw=2, label=f'CLT Normal N({pop_mean:.2f}, {clt_std:.4f}²)')
ax.set_title('CLT: Distribution of Sample Means')
ax.legend()
plt.tight_layout(); plt.show()

# KS test to confirm approximate normality
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

> *In 2–3 sentences: Why does this result matter for statistics and machine learning? What assumption does the CLT justify in many classical statistical tests?*

### CLT Reflection

The CLT tells us that no matter what shape the original population has (here, exponential — skewed), the distribution of **sample means** converges to a normal distribution as sample size grows. This is foundational for statistics because it justifies using normal-distribution-based tests (t-tests, z-tests, confidence intervals) even on non-normal data, as long as the sample size is reasonably large (n ≥ 30 is a common rule of thumb). In ML, it underpins why we trust aggregate metrics (mean loss, mean accuracy) to follow predictable distributions, enabling principled hypothesis testing about model performance.
